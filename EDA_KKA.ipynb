#**Latihan 3**

```python
df = pd.read_csv('/data_kantin - data_kantin.csv')

print(df.head())
print()
print(df.info())
print()
print(df.describe())
print()
print(df.shape)
```

## **Analisis 3**
Kolom yang jumlah non-null-nya lebih sedikit dari jumlah baris adalah menu (107) dan terjual (104). Artinya, kedua kolom tersebut memiliki beberapa data yang kosong.

#**Latihan 4**

```python
print(df.isnull().sum())

df['terjual'] = df['terjual'].fillna(0)
df = df.dropna(subset=['menu'])
```

## **Analisis 4**
Kolom terjual diisi 0 supaya tetap bisa dihitung. Sedangkan menu yang kosong dihapus karena nama menu diperlukan untuk mengetahui data yang sedang dianalisis.

#**Latihan 5**

```python
print(df.duplicated().sum())
df = df.drop_duplicates()

df['harga'] = df['harga'].astype(int)
print(df.dtypes)
```

##**Analisis 5**
Jumlah baris berkurang jika terdapat data yang sama. Tipe data perlu diperiksa supaya perhitungan dan pengolahan data tidak mengalami masalah.

#**Latihan 6**

```python
laris = df[df['terjual'] > 20]
urut = df.sort_values(by='terjual', ascending=False)

df['total_pendapatan'] = df['harga'] * df['terjual']

ringkasan = df.groupby('menu')['total_pendapatan'].sum()
print(ringkasan)
```

## **Analisis 6**
Menu yang menghasilkan total pendapatan tertinggi adalah Nasi Goreng, yaitu Rp3.924.000. Data ini bisa membantu kantin mengetahui menu yang menghasilkan pendapatan paling besar.

**Rencana Manipulasi Data**

```python
import pandas as pd

df = pd.read_csv('/EDA-KKA-DATASET.csv')

print("=== 5 DATA PERTAMA ===")
print(df.head())

print("\n=== INFO DATA ===")
print(df.info())

print("\n=== STATISTIK DATA ===")
print(df.describe())

print("\n=== UKURAN DATA ===")
print(df.shape)

# P4 - Cleaning
print("\n=== MISSING VALUE SEBELUM CLEANING ===")
print(df.isnull().sum())

df['Terjual'] = df['Terjual'].fillna(0)
df = df.dropna(subset=['Produk'])

print("\n=== MISSING VALUE SETELAH CLEANING ===")
print(df.isnull().sum())

print("\n=== DATA DUPLIKAT ===")
print("Jumlah duplikat:", df.duplicated().sum())

df = df.drop_duplicates()

print("Jumlah baris setelah cleaning:", len(df))

# P5 - Manipulation
print("\n=== FILTER: TERJUAL > 20 ===")
laris = df[df['Terjual'] > 20]
print(laris)

print("\n=== SORTING: TERJUAL TERBANYAK ===")
urut = df.sort_values(by='Terjual', ascending=False)
print(urut)

# Kolom turunan
df['total_pendapatan'] = df['Harga'] * df['Terjual']

print("\n=== DATA DENGAN TOTAL PENDAPATAN ===")
print(df)

# Groupby
print("\n=== TOTAL PENDAPATAN PER PRODUK ===")
ringkasan = df.groupby('Produk')['total_pendapatan'].sum()
print(ringkasan)
```
